# Partie 1 : Perceptron

## 1. Définition des variables x1 et x2

Les variables d'entrée du perceptron, x1 et x2, prennent les valeurs 0 ou 1.

```python
x1 = 0
x2 = 1 
```

## 2. Fonction de calcul de s

Voici une fonction permettant de calculer cette expression :

$$
s = w_0 + x_1 \cdot w_1 + x_2 \cdot w_2
$$

```python
def calcul_s(x1, x2, w0, w1, w2):
    """
    Calcule la somme pondérée s pour le perceptron.
    """
    s = w0 + x1 * w1 + x2 * w2
    return s
```

## 3. Fonction d'activation
Nous créons une deuxième fonction qui prend en entrée $s$ et retourne $y$ selon la règle suivante :

- Si $s \geq 0$, alors $y = 1$
- Si $s < 0$, alors $y = 0$

```python
def activation(s):
    """
    Fonction d'activation du perceptron (seuil à 0).
    """
    if s >= 0:
        return 1
    else:
        return 0
```

## Exemple d'utilisation
Nous allons maintenant utiliser les fonctions précédentes avec des valeurs d'exemple pour les poids et les entrées.

```python
#Poids du perceptron
w0 = -0.5
w1 = 1.0
w2 = 1.0

#entrées
x1 = 0
x2 = 1

#calcul de la sosrtie
s = calcul_s(x1, x2, w0, w1, w2)

#calcul de la sortie activée
y = activation(s)

print(f"s = {s}")
print(f"y = {y}")

s = calcul_s(x1,x1, w0, w0, w2)
y = activation(s)
print(f"s2 = {s}")
print(f"y2 = {y}")
```

## 4. Validation des opérateurs logiques (OU et ET)

Nous allons maintenant tester le perceptron pour tous les cas des tables de vérité des opérateurs logiques OU (OR) et ET (AND), et vérifier les sorties obtenues pour chaque combinaison d'entrées.

```python
import pandas as pd

#table de vérité pour OU (OR)
poids_ou = {'w0': -0.5, 'w1': 1.0, 'w2': 1.0}
#table de vérité pour ET (AND)
poids_et = {'w0': -1.5, 'w1': 1.0, 'w2': 1.0}

combinaisons = [(0,0), (0,1), (1,0), (1,1)]

#test pour l'opérateur OU
resultats_ou = []
for x1, x2 in combinaisons:
    s = calcul_s(x1, x2, **poids_ou)
    y = activation(s)
    resultats_ou.append({'x1': x1, 'x2': x2, 's': s, 'y': y})
df_ou = pd.DataFrame(resultats_ou)

#test pour l'opérateur ET
resultats_et = []
for x1, x2 in combinaisons:
    s = calcul_s(x1, x2, **poids_et)
    y = activation(s)
    resultats_et.append({'x1': x1, 'x2': x2, 's': s, 'y': y})
df_et = pd.DataFrame(resultats_et)

print('Table de vérité OU (OR)')
print(df_ou)
print('\nTable de vérité ET (AND)')
print(df_et)
```

**Est‐ce qu’on peut obtenir des poids satisfaisant les valeurs des opérateurs logiques ET/OU ?**


Oui car les deux fonctions logiques sont linéairement séparables : il existe une droite qui sépare parfaitement les sorties 0 et 1 dans l’espace des entrées.

**Est‐ce qu’on peut représenter à l’aide d’un perceptron l’opérateur OU‐Exclusif (XOR) ? Oui, Non. Pourquoi ?**

Non car contrairement aux opérateurs logiques ET/OU, les portes XOR/XNOR ne sont pas linéarement séparable. Il faut donc deux couches pour répondre à ce problème.
